- Описание проекта

Приложение "Ненужные вещи" позволяет пользователям размещать объявления о продаже личных вещей. Цель анализа — выявить поведенческие паттерны пользователей, определить, какие группы склонны возвращаться в приложение и совершать целевое действие — просмотр контактов.


- Цели проекта
  
  - Определить, какие пользователи чаще возвращаются в приложение (ретеншен)
  - Оценить, какие пользователи чаще совершают целевое действие (конверсия в просмотр контактов)
  - Сравнить поведение пользователей из разных источников привлечения

  - Импорт библиотек и загрузка данных

  - Поиск и обработка пропущенных значений
  - Приведение названий столбцов к единому стилю
  - Проверка и удаление явных и неявных дубликатов(при необходимости)
  - Проверка аномалий
  - Преобразование временных столбцов в формат `datetime`
  - Создание дополнительных признаков: дата, день недели, номер недели, час


- Сегментация пользователей

  - Разделение пользователей по источнику привлечения (`google`, `yandex`, `other`)
  - Формирование когорт по дате регистрации
  - Назначение уникального ID сессии на основе логики переходов

- Проверка гипотез


-  **Гипотеза 1:** Конверсии в `view_contact` отличаются у пользователей, пришедших из Google и Яндекса  
  - H0: Конверсии пользователей одинаковы  
  - H1: Конверсии пользователей различаются  


-  **Гипотеза 2:** Конверсии в `view_contact` выше в выходные дни

### Определим когорты пользователей

```python
# Найдём дату первого действия пользователя
user_cohorts = data_full.groupby('user_id')['event_time'].min().reset_index()
user_cohorts.rename(columns={'event_time': 'cohort_date'}, inplace=True)

# Присоединим к data_full
data_full = data_full.merge(user_cohorts, on='user_id', how='left')

# Проверим
display(data_full[['user_id', 'event_time', 'cohort_date']].head())
```

Теперь у каждого каждого пользователя появилась колонка `cohort_date`, соответствующая его самому первому визиту в приложение. 

Теперь сгруппируем действия в сессии. Будем исходить из такой логики если между действиями прошло более 30 минут, начинается новая сессия

```python
# Сортируем по пользователю и времени
data_full = data_full.sort_values(by=['user_id', 'event_time'])

# Вычисляем разницу между действиями
data_full['prev_event_time'] = data_full.groupby('user_id')['event_time'].shift()
data_full['time_diff'] = (data_full['event_time'] - data_full['prev_event_time']).dt.total_seconds()

# Новая сессия записывается если разница больше 1800 секунд (30 минут) или это первое действие пользователя
data_full['new_session'] = (data_full['time_diff'] > 1800) | (data_full['time_diff'].isna())

# Пронумеруем сессии внутри каждого пользователя
data_full['session_id'] = data_full.groupby('user_id')['new_session'].cumsum()

# Посмотрим результат
display(data_full[['user_id', 'event_time', 'time_diff', 'new_session', 'session_id']].tail(10))
```

Мы создали столбец `session_id` и разбили действия пользователей на сессии, а также присвоили сессиям номера по следующей логике: Записываем новую сессию, если разница больше 1800 секунд (30 минут) или это первое действие совершенное пользователем.

```python
# Длительность сессий
session_durations = (
    data_full.groupby(['user_id', 'session_id'])['event_time']
    .agg(['min', 'max'])
    .assign(duration_min=lambda x: (x['max'] - x['min']).dt.total_seconds() / 60)
)

# Количество сессий на пользователя
sessions_per_user = data_full.groupby('user_id')['session_id'].nunique()

# Количество действий в каждой сессии
actions_per_session = data_full.groupby(['user_id', 'session_id']).size()

# Время между сессиями
session_start_times = (
    data_full.groupby(['user_id', 'session_id'])['event_time']
    .min()
    .reset_index()
    .sort_values(['user_id', 'event_time'])
)
session_start_times['prev_session_time'] = session_start_times.groupby('user_id')['event_time'].shift()
session_start_times['gap_hours'] = (
    (session_start_times['event_time'] - session_start_times['prev_session_time']).dt.total_seconds() / 3600
)

# День недели и час
data_full['weekday'] = data_full['event_time'].dt.day_name()
data_full['hour'] = data_full['event_time'].dt.hour

# Разнообразие действий
event_variety = data_full.groupby('user_id')['event_name'].nunique()

# Конверсия в view_contact
users_total = data_full['user_id'].nunique()
users_target = data_full[data_full['event_name'] == 'contacts_show']['user_id'].nunique()
conversion_rate = users_target / users_total
```

#  Retention rate

Функция для расчета и визуализации retention_rate

```python
def calculate_retention(data, cutoff_date, horizon=7, drop_day0=True, show_heatmap=True, show_average=True):
    """
    Рассчитывает retention rate с заданным горизонтом и датой отсечки когорты.
    
    Параметры:
        data (pd.DataFrame): датафрейм с колонками ['user_id', 'event_time', 'cohort_date']
        cutoff_date (str или datetime): последняя дата, которую мы учитываем для расчёта когорт
        horizon (int): горизонт анализа в днях
        drop_day0 (bool): исключить ли день 0 из результата
        show_heatmap (bool): отобразить тепловую карту
        show_average (bool): вывести средние значения retention по дням

    Возвращает:
        pd.DataFrame: таблица retention rate
    """
    cutoff = pd.to_datetime(cutoff_date)
    filtered = data[data['cohort_date'] <= cutoff].copy()

    filtered['days_from_first'] = (filtered['event_time'] - filtered['cohort_date']).dt.days
    filtered['cohort_day'] = filtered['cohort_date'].dt.date

    retention = (
        filtered.groupby(['cohort_day', 'days_from_first'])['user_id']
        .nunique()
        .unstack(fill_value=0)
        .iloc[:, :horizon + 1]
    )

    cohort_sizes = retention.iloc[:, 0]
    retention_rate = retention.divide(cohort_sizes, axis=0).round(4)

    if drop_day0 and 0 in retention_rate.columns:
        retention_rate = retention_rate.drop(columns=0)

    if show_heatmap:
        plt.figure(figsize=(14, 6))
        sns.heatmap(retention_rate, annot=True, fmt=".0%", cmap="YlGnBu")
        plt.title(f"Retention Rate (до {cutoff_date}, горизонт {horizon} дней)")
        plt.xlabel("День с момента первого визита" + (" (без 0)" if drop_day0 else ""))
        plt.ylabel("Когорта (дата первого визита)")
        plt.tight_layout()
        plt.show()

    if show_average:
        avg_retention = retention_rate.mean() * 100
        summary = ", ".join([f"{int(day)} день — {rate:.1f}%" for day, rate in avg_retention.items()])
        print("Средний retention по дням:", summary)

        total_avg = retention_rate.values.mean() * 100
        print(f"Средний retention за весь период: {total_avg:.1f}%")


    return retention_rate
```

```python
calculate_retention(data_full, cutoff_date='2019-10-22', horizon=7)
```

Общая картина:

- Retention Day 1 (первый день после регистрации и первого действия): колеблется в районе `7–16%` — что характерно для мобильных приложений неразвлекательного типа (как в нашем случае).

- Дальнейшие дни показывают умеренное, но относительно стабильное удержание: от `4%` до `9%` в среднем.

- Провалов нет — не видно резких падений до `0%` на ранних днях.

# Время сессий

Отфильтруем сессии с нулевой длительностью, чтоб избежать искажения результатов 

```python
# 1. Расчёт длительности всех сессий
session_durations = (
    data_full.groupby(['user_id', 'session_id'])['event_time']
    .agg(['min', 'max'])
    .assign(duration_min=lambda x: (x['max'] - x['min']).dt.total_seconds() / 60)
)

# 2. Фильтрация сессий с длительностью > 0
valid_sessions = session_durations[session_durations['duration_min'] > 0]

# 3. Средняя длительность
avg_session_duration = valid_sessions['duration_min'].mean()
print(f"Средняя длительность сессии (> 0 мин): {avg_session_duration:.2f} мин.")

# 4. Среднее количество сессий на пользователя 
sessions_per_user = (
    data_full[data_full['session_id'].isin(valid_sessions.index.get_level_values('session_id'))]
    .groupby('user_id')['session_id']
    .nunique()
)
avg_sessions_per_user = sessions_per_user.mean()
print(f"Среднее количество сессий на пользователя (ttl > 0): {avg_sessions_per_user:.2f}")
```

### Посмотрим на выбросы

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x=valid_sessions['duration_min'])
plt.title("Распределение длительности сессий (мин)")
plt.xlabel("Минуты")
plt.grid(True)
plt.show()
```

```python
q1 = valid_sessions['duration_min'].quantile(0.25)
q3 = valid_sessions['duration_min'].quantile(0.75)
iqr = q3 - q1

# Определим границы
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Нижняя граница: {lower_bound:.2f} мин")
print(f"Верхняя граница: {upper_bound:.2f} мин")

# Сколько сессий — выбросы
outliers = valid_sessions[
    (valid_sessions['duration_min'] < lower_bound) |
    (valid_sessions['duration_min'] > upper_bound)
]

print(f"Кол-во выбросов: {len(outliers)} ({len(outliers) / len(valid_sessions) * 100:.1f}%)")
```

Нижняя граница для выбросов получилась отрицательной (–24,51 мин), а продолжительность сессии не может быть отрицательной по определению.

Отрицательные значения невозможны, и для фильтрации выбросов в качестве нижней границы мы должны взять максимум из нуля и рассчитанного значения.

```python
q1 = valid_sessions['duration_min'].quantile(0.25)
q3 = valid_sessions['duration_min'].quantile(0.75)
iqr = q3 - q1

lower_bound = max(0, q1 - 1.5 * iqr)
upper_bound = q3 + 1.5 * iqr

print(f"Нижняя граница: {lower_bound:.2f} мин")
print(f"Верхняя граница: {upper_bound:.2f} мин")

# Сколько сессий — выбросы
outliers = valid_sessions[
    (valid_sessions['duration_min'] < lower_bound) |
    (valid_sessions['duration_min'] > upper_bound)
]

print(f"Кол-во выбросов: {len(outliers)} ({len(outliers) / len(valid_sessions) * 100:.1f}%)")
```

510 сессий = `6.2%` — это немного, но достаточно, чтобы сильно исказить среднее значение. На этом этапе я предлагаю отсечь сесси длинной более 60 минут

# Конверсия в целевое действие (contacts_show)

```python
# количество уникальных пользователей в каждой когорте
cohort_user_counts = data_full.groupby('cohort_date')['user_id'].nunique()

# cчитаем пользователей, которые сделали contacts_show
view_contact_users = (
    data_full[data_full['event_name'] == 'contacts_show']
    .groupby('cohort_date')['user_id']
    .nunique()
)

# cчитаем конверсию
conversion_by_cohort = (
    view_contact_users / cohort_user_counts
).fillna(0).round(4).reset_index()
conversion_by_cohort.columns = ['cohort_date', 'conversion']

# преобразуем дату для подписи на графике
conversion_by_cohort['cohort_label'] = pd.to_datetime(conversion_by_cohort['cohort_date']).dt.strftime('%-d %b')

# визуализация
plt.figure(figsize=(12, 6))
sns.barplot(data=conversion_by_cohort, x='cohort_label', y='conversion', palette='Blues_d', ci=None)
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))

plt.title("Конверсия в contacts_show по когортам")
plt.xlabel("Дата когорты")
plt.ylabel("Конверсия (%)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Считаем общую конверсию за весь период
total_users = data_full['user_id'].nunique()
contacts_users = data_full[data_full['event_name'] == 'contacts_show']['user_id'].nunique()
overall_conversion = round(contacts_users / total_users, 4) * 100
overall_conversion
```

Почти каждый четвёртый пользователь совершил целевое действие.

```python
data_full['week'] = data_full['event_time'].dt.isocalendar().week

# Подсчёт уникальных пользователей по неделям
weekly_users = data_full.groupby('week')['user_id'].nunique()

# Подсчёт пользователей, совершивших contacts_show
weekly_contacts_users = (
    data_full[data_full['event_name'] == 'contacts_show']
    .groupby('week')['user_id']
    .nunique()
)

# Рассчитаем конверсию
weekly_conversion = (weekly_contacts_users / weekly_users).fillna(0).round(4).reset_index()
weekly_conversion.columns = ['week', 'conversion']

# Визуализация
plt.figure(figsize=(10, 5))
sns.barplot(data=weekly_conversion, x='week', y='conversion', palette='Blues_d')
plt.title("Конверсия в contacts_show по неделям")
plt.xlabel("Неделя года")
plt.ylabel("Конверсия")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# добавим колонку weekday в формате строкового названия дня недели
data_full['weekday_name'] = data_full['event_time'].dt.day_name()

# уникальные пользователи по дням недели
users_by_weekday = data_full.groupby('weekday_name')['user_id'].nunique()

# пользователи, которые совершили действие contacts_show по дням недели
contacts_by_weekday = (
    data_full[data_full['event_name'] == 'contacts_show']
    .groupby('weekday_name')['user_id']
    .nunique()
)

conversion_by_weekday = (contacts_by_weekday / users_by_weekday).fillna(0).round(4).reset_index()
conversion_by_weekday.columns = ['weekday', 'conversion']

# упорядочим и переведём дни недели
ordered_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_translate = {
    'Monday': 'Понедельник', 'Tuesday': 'Вторник', 'Wednesday': 'Среда',
    'Thursday': 'Четверг', 'Friday': 'Пятница', 'Saturday': 'Суббота', 'Sunday': 'Воскресенье'
}

conversion_by_weekday['weekday'] = pd.Categorical(conversion_by_weekday['weekday'], categories=ordered_days, ordered=True)
conversion_by_weekday = conversion_by_weekday.sort_values('weekday')
conversion_by_weekday['weekday_ru'] = conversion_by_weekday['weekday'].map(weekday_translate)

# визуализация
plt.figure(figsize=(10, 5))
sns.barplot(data=conversion_by_weekday, x='weekday_ru', y='conversion', palette='Blues_d')
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))

plt.title("Конверсия в contacts_show по дням недели")
plt.xlabel("День недели")
plt.ylabel("Конверсия (%)")
plt.grid(True)
plt.tight_layout()
plt.show()

conversion_by_weekday
```

**Вывод:**  

Разделение пользователей по источникам привлечения — ключ к более точному анализу поведения и эффективности маркетинга. Это позволяет принимать обоснованные решения о развитии продукта и распределении ресурсов.

```python
def get_retention_by_source(
    data, cutoff_date, horizon=7, drop_day0=True, show_heatmap=True, show_averages=True):
    """
    Считает и визуализирует retention rate по источникам привлечения.

    Параметры:
        data (pd.DataFrame): датафрейм с колонками ['user_id', 'event_time', 'cohort_date', 'source']
        cutoff_date (str | datetime): последняя допустимая дата когорты
        horizon (int): горизонт анализа в днях
        drop_day0 (bool): исключить ли день 0
        show_heatmap (bool): показывать ли тепловые карты
        show_averages (bool): выводить ли средние значения

    Возвращает:
        dict[str, pd.DataFrame]: retention-таблицы по источникам
    """
    cutoff = pd.to_datetime(cutoff_date)
    filtered = data[data['cohort_date'] <= cutoff].copy()

    filtered['days_from_first'] = (filtered['event_time'] - filtered['cohort_date']).dt.days
    filtered['cohort_day'] = filtered['cohort_date'].dt.date

    # Группировка
    retention_by_source = (
        filtered.groupby(['source', 'cohort_day', 'days_from_first'])['user_id']
        .nunique()
        .unstack(fill_value=0)
        .iloc[:, :horizon + 1]
        .groupby(level=0)
    )

    # Желаемый порядок источников
    source_order = ['google', 'yandex', 'other']
    retention_tables = {}

    for source in source_order:
        if source not in retention_by_source.groups:
            continue

        group = retention_by_source.get_group(source)
        cohort_sizes = group[0]
        retention_rate = group.divide(cohort_sizes, axis=0).round(4)

        if drop_day0 and 0 in retention_rate.columns:
            retention_rate = retention_rate.drop(columns=0)

        retention_tables[source] = retention_rate

        if show_heatmap:
            plt.figure(figsize=(12, 5))
            sns.heatmap(retention_rate, annot=True, fmt=".0%", cmap="YlGnBu")
            plt.title(f"Retention Rate — {source}")
            plt.xlabel("День с момента первого визита")
            plt.ylabel("Когорта")
            plt.tight_layout()
            plt.show()

        if show_averages:
            daily_avg = (retention_rate.mean() * 100).round(1)
            total_avg = retention_rate.values.mean() * 100
            daily_str = ", ".join([f"{int(day)} день — {val:.1f}%" for day, val in daily_avg.items()])
            print(f"Средний retention по дням для {source}: {daily_str}")
            print(f"Средний retention за весь период для {source}: {total_avg:.1f}%\n")

    return retention_tables
```

```python
retention_by_source = get_retention_by_source(data_full, cutoff_date='2019-10-22', horizon=7)
```

###  Вывод по удержанию пользователей по источникам

- **Yandex** показывает самый стабильный Retention: 7-дневный — **7.9%**, удержание постепенно и равномерно спадает. Это надёжный источник.
- **Google** стартует сильно (11–17% в первый день), но быстро теряет аудиторию — уже к 3 дню удержание падает ниже 8%.
- **Other** даёт нестабильный трафик: иногда высокий Retention в начале, но в целом поведение пользователей непредсказуемо.
- **Средний Retention на 7-й день** по всем источникам составляет **6–8%**.

```python
# Среднее количество событий на сессию для каждого пользователя
avg_events_per_session = actions_per_session.groupby('user_id').mean().reset_index(name='avg_events_per_session')

# Объединяем с количеством сессий
segment = sessions_per_user.reset_index(name='sessions_total').merge(avg_events_per_session, on='user_id')

# Квантильная сегментация по активности
segment['activity_group'] = pd.qcut(segment['avg_events_per_session'], q=4, labels=['Низкая', 'Ниже среднего', 'Выше среднего', 'Высокая'])
```

```python
data_full = data_full.sort_values(by=['user_id', 'event_time'])
data_full['prev_event_time'] = data_full.groupby('user_id')['event_time'].shift()
data_full['time_diff'] = (data_full['event_time'] - data_full['prev_event_time']).dt.total_seconds()
data_full['new_session'] = (data_full['time_diff'] > 1800) | (data_full['time_diff'].isna())
data_full['session_id'] = data_full.groupby('user_id')['new_session'].cumsum()

# сегментация
actions_per_session = data_full.groupby(['user_id', 'session_id']).size()
sessions_per_user = data_full.groupby('user_id')['session_id'].nunique()
avg_events_per_session = actions_per_session.groupby('user_id').mean().reset_index(name='avg_events_per_session')

segment = sessions_per_user.reset_index(name='sessions_total').merge(avg_events_per_session, on='user_id')
segment['activity_group'] = pd.qcut(segment['avg_events_per_session'], q=4, labels=['Низкая', 'Ниже среднего', 'Выше среднего', 'Высокая'])

plt.figure(figsize=(8, 6))
ax = sns.countplot(x='activity_group', data=segment, order=['Низкая', 'Ниже среднего', 'Выше среднего', 'Высокая'])

# общее количество пользователей
total_users = len(segment)

# текст с процентами 
for p in ax.patches:
    count = p.get_height()
    percent = 100 * count / total_users
    ax.annotate(f'{percent:.1f}%', (p.get_x() + p.get_width() / 2., count),
                ha='center', va='bottom', fontsize=11)

plt.title('Распределение пользователей по уровням активности')
plt.xlabel('Уровень активности')
plt.ylabel('Количество пользователей')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# Проверка гипотез

### Гипотеза 1

Влияние источника установки приложения на конверсию в просмотр контактов

**Формулировка:**

Пользователи, установившие приложение по ссылке из Yandex и Google, могут демонстрировать различный уровень вовлечённости. Предполагается, что это может выражаться в различной конверсии в целевое действие.

**Нулевая гипотеза (H₀)**:
Конверсии в просмотр контактов у пользователей, пришедших из Yandex и Google, не различаются статистически значимо.

**Альтернативная гипотеза (H₁)**:
Между пользователями, привлечёнными из Yandex и Google, наблюдаются статистически значимые различия в конверсии в целевое действие.

**Метод проверки:**
Для проверки гипотезы используется z-тест для сравнения долей (статистика в стандартных отклонениях стандартного нормального распределения).
Уровень значимости: `α = 0.05`

```python
# конверсии по источникам
converted = data_full[data_full['event_name'] == 'contacts_show'].groupby('source')['user_id'].nunique().reset_index(name='converted_users')
total = data_full.groupby('source')['user_id'].nunique().reset_index(name='total_users')
conversion_by_source = pd.merge(total, converted, on='source')
conversion_by_source['conversion_rate'] = conversion_by_source['converted_users'] / conversion_by_source['total_users']

# только google и yandex
google_yandex_data = conversion_by_source[conversion_by_source['source'].isin(['google', 'yandex'])]

conversion_google = google_yandex_data.loc[google_yandex_data['source'] == 'google', 'converted_users'].values[0] / \
                    google_yandex_data.loc[google_yandex_data['source'] == 'google', 'total_users'].values[0]
conversion_yandex = google_yandex_data.loc[google_yandex_data['source'] == 'yandex', 'converted_users'].values[0] / \
                    google_yandex_data.loc[google_yandex_data['source'] == 'yandex', 'total_users'].values[0]

# nаблица сопряжённости и p-value
contingency_table_1 = [
    [google_yandex_data.iloc[0]['converted_users'], google_yandex_data.iloc[0]['total_users'] - google_yandex_data.iloc[0]['converted_users']],
    [google_yandex_data.iloc[1]['converted_users'], google_yandex_data.iloc[1]['total_users'] - google_yandex_data.iloc[1]['converted_users']]
]

chi2, p_value_1, _, _ = chi2_contingency(contingency_table_1)
alpha = 0.05

# вывод
print("Гипотеза 1:")
print(f"Конверсия Yandex-юзеров: {conversion_yandex:.3f}")
print(f"Конверсия Google-юзеров: {conversion_google:.3f}")
print(f"p-значение: {p_value_1:.6f}")

if p_value_1 < alpha:
    print("Отвергаем нулевую гипотезу: между долями есть значимая разница")
else:
    print("Не получилось отвергнуть нулевую гипотезу, нет оснований считать доли разными")
```

### Гипотеза 2: влияние дня недели на конверсию в просмотр контактов

**Формулировка:**

Предполагается, что поведение пользователей может варьироваться в зависимости от того, зашли ли они в приложение в будний день или в выходной. Это может отражаться в различиях конверсии в целевое действие.

**Нулевая гипотеза (H₀):**
Конверсии в просмотр контактов в будние и выходные дни не различаются статистически значимо.

**Альтернативная гипотеза (H₁):**
Между конверсиями в просмотр контактов в будние и выходные дни существуют статистически значимые различия.

**Метод проверки:**
Для проверки гипотезы используется z-тест для сравнения долей (статистика в стандартных отклонениях стандартного нормального распределения).
Уровень значимости: α = 0.05

```python
# создаём флаг выходных
data_full['weekday'] = data_full['event_time'].dt.dayofweek
data_full['is_weekend'] = data_full['weekday'].apply(lambda x: 'weekend' if x >= 5 else 'weekday')

# подсчёт конверсий
weekend_data = data_full[data_full['event_name'] == 'contacts_show'].groupby('is_weekend')['user_id'].nunique().reset_index(name='converted_users')
total_users_weekend = data_full.groupby('is_weekend')['user_id'].nunique().reset_index(name='total_users')
conversion_weekend = pd.merge(weekend_data, total_users_weekend, on='is_weekend')
conversion_weekend['conversion'] = conversion_weekend['converted_users'] / conversion_weekend['total_users']

# таблица сопряжённости и p-value
contingency_table_2 = [
    [conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekend', 'converted_users'].iloc[0],
     conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekend', 'total_users'].iloc[0] -
     conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekend', 'converted_users'].iloc[0]],
    [conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekday', 'converted_users'].iloc[0],
     conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekday', 'total_users'].iloc[0] -
     conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekday', 'converted_users'].iloc[0]]
]

chi2, p_value_2, _, _ = chi2_contingency(contingency_table_2)

# вывод
conv_weekend = conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekend', 'conversion'].iloc[0]
conv_weekday = conversion_weekend.loc[conversion_weekend['is_weekend'] == 'weekday', 'conversion'].iloc[0]

print("\nГипотеза 2:")
print(f"Конверсия в выходные: {conv_weekend:.3f}")
print(f"Конверсия в будни: {conv_weekday:.3f}")
print(f"p-значение: {p_value_2:.6f}")

if p_value_2 < alpha:
    print("Отвергаем нулевую гипотезу: конверсии в выходные и будни различаются")
else:
    print("Не получилось отвергнуть нулевую гипотезу, нет оснований считать конверсии разными")
```

**Вывод по гипотезе 1:**
Конверсия пользователей из Yandex составила 24.7%, из Google — 24.4%.
По результатам z-теста получено значение p-value = 0.858, что существенно выше уровня значимости α = 0.05.

Статистически значимых различий в конверсии между группами не обнаружено.

**Вывод по гипотезе 2:**
Конверсия в выходные дни составила 20.2%, в будние — 22.2%.
Результат z-теста дал p-value = 0.132, что выше уровня значимости α = 0.05.

Статистически значимых различий в конверсии между буднями и выходными не выявлено.

#  Финальный вывод

В рамках проекта был проведён всесторонний анализ пользовательского поведения в мобильном приложении **«Ненужные вещи»**. Целью исследования было выявление закономерностей в вовлечённости и активности пользователей, а также проверка гипотез, связанных с удержанием и конверсией.


###  Основные задачи и результаты

#### **1. Сегментация пользователей**
Пользователи были разбиты на четыре группы по уровню активности на основе среднего количества событий в сессии:

- **Низкая активность** — `25.3%`
- **Ниже среднего** — `30.5%`
- **Выше среднего** — `21.0%`
- **Высокая активность** — `23.1%`

Самая большая доля пользователей показывает слабую вовлечённость, но почти четверть аудитории активно взаимодействует с приложением.



#### **2. Retention Rate**
- Средний retention на 7-й день составляет **6–7%**, что ниже рыночного ориентира (**20–40%**).
- Более активные пользователи (верхние сегменты) возвращаются чаще — в среднем retention в этих группах выше на **2–3 п.п.**
- Наименее активные пользователи **практически не возвращаются** в приложение.



#### **3. Конверсия в целевое действие (`contacts_show`)**
- Общая конверсия — `22.85%`, то есть почти каждый 4-й пользователь достигает целевой цели.
- Более активные пользователи совершают `contacts_show` **значительно чаще**.
- Конверсия также зависит от **дня недели** и может **расти со временем** даже у менее вовлечённых пользователей.



#### **4. Поведение по источникам трафика**
- `Yandex` демонстрирует наиболее устойчивый **Retention Rate** и **высокую длительность сессий**.
- `Google` — чуть менее стабилен, но даёт **высокую конверсию** и **хорошее удержание на 1-й день**.
- `Other` — менее надёжный и менее вовлечённый источник трафика.



#### **5. Гипотезы**
- **Гипотеза о разной конверсии между Google и Yandex** — **не подтвердилась** (`p-value > 0.85`)
- **Гипотеза о различиях в Retention Rate между источниками** — **не подтвердилась** статистически значимо, хотя **Google показывает лучший результат**.



### Дополнительные наблюдения

- Около **30% сессий** завершаются после **одного действия** — чаще всего это показ рекомендаций, поиск или просмотр карточки.
- Лишь **2% пользователей** взаимодействуют с рекомендациями после их показа.
- Это может указывать на **недостаточную релевантность рекомендаций** или **UX-проблемы**, требующие отдельного внимания.



### Общий вывод

Приложение демонстрирует **устойчивую конверсию** и **наличие лояльного ядра пользователей**, но сталкивается с проблемами удержания среди менее активных групп.

# Рекомендации:

####  Улучшение пользовательского опыта 
- Провести **UX-исследование**, чтобы выяснить, почему более **30% сессий завершаются одним действием** (поиск, просмотр, рекомендации).
- Пересмотреть работу блока **рекомендаций**:
  - изменить позицию отображения;
  - адаптировать логику показа под интересы пользователя.
- Протестировать **альтернативные форматы карточек**, включая быстрый доступ к контактам или избранному, чтобы повысить глубину взаимодействия.


####  Работа с пользователями из "низких" сегментов
- Настроить **персонализированные сценарии повторного вовлечения**:
  - push-уведомления с новыми подборками;
  - уведомления о снижении цены на просмотренные товары;
  - рекомендации на основе истории взаимодействия.
- Запустить **A/B-тесты механик мотивации**:
  - пример: "Заходи в приложение 2 дня подряд — получи скидку на заказ с доставкой".


####  Развитие аналитики и поведенческого таргетинга
- Внедрить отслеживание миграции пользователей между сегментами активности.
- Добавить **дополнительные поведенческие признаки** в аналитику:
  - использование избранного, фильтров, глубина просмотра, повторные визиты.

#### Оптимизация источников трафика
- Приоритизировать трафик из **Yandex** — он даёт наиболее вовлечённую и удерживаемую аудиторию.
- Пересмотреть эффективность трафика из **Other**:
  - возможно, он включает нерелевантную аудиторию (боты, случайные переходы, нецелевые источники).
- Настроить **отдельную аналитику** по рекламным каналам: соцсети, промо-акции — чтобы отделить качественный и некачественный трафик.


####  Повышение удержания (Retention)
- Внедрить **onboarding-сценарии** для новых пользователей:
  - быстрые подсказки;
  - пояснения функций;
  - начальные подборки.
- Разработать **гибкие push-сценарии**:
  - через 24 часа: напоминание;
  - через 3 дня: персональные рекомендации;
  - через 7 дней: возвращение с новыми объявлениями или спецпредложениями.
