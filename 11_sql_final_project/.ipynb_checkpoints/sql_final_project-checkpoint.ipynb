```python
# Создаём функцию для универсальной загрузки таблиц
def get_table(table_name, engine):
    """Загружает таблицу из базы данных в DataFrame по названию"""
    return pd.read_sql(f"SELECT * FROM {table_name}", con=engine)

# Список нужных таблиц
table_names = ['books', 'authors', 'publishers', 'ratings', 'reviews']
# Загрузка всех таблиц в словарь (по ключу - название таблицы)
tables = {name: get_table(name, engine) for name in table_names}
```

```python
def run_query(sql):
    return pd.read_sql(sql, con=engine)
```

## Посчитайте количество книг, выпущенных после 1 января 2000 года

```python
books_cnt = run_query("""
    SELECT COUNT(book_id) AS books_cnt 
    FROM books 
    WHERE publication_date >= '2000-01-02'
""")
print(books_cnt)
```

**Вывод**: После `01.01.2000` было выпущено `819` книг

## Для каждой книги посчитайте количество обзоров и среднюю оценку

```python
ratings_reviews_query = """
SELECT
    books.title AS title,
    books.book_id AS id,
    AVG(ratings.rating) AS avg_rating,
    COUNT(DISTINCT reviews.text) AS cnt_reviews
FROM
    books
INNER JOIN ratings ON ratings.book_id = books.book_id
LEFT JOIN reviews ON reviews.book_id = books.book_id
GROUP BY
    books.title,
    books.book_id
ORDER BY 
    cnt_reviews DESC
"""

ratings_reviews = run_query(ratings_reviews_query)
ratings_reviews
```

Для семи книг отсутствуют как рейтинг, так и обзор. Максимальное количество обзоров зафиксировано у книги «`Twilight (Twilight 1`)(Сумерки)». Достаточно большое число книг имеют оценку 5, однако для каждой из них представлено лишь по два обзора.

## Определите издательство, которое издало наибольшее число книг толще 50 страниц 

```python
book_publisher_query = """
SELECT
     publishers.publisher AS name_publisher,
     COUNT(books.book_id) AS cnt_of_published_books
FROM
    publishers
INNER JOIN books ON books.publisher_id = publishers.publisher_id
WHERE
    books.num_pages > 50
GROUP BY
    name_publisher
ORDER BY 
    cnt_of_published_books DESC
LIMIT 20
"""

book_publisher = run_query(book_publisher_query)
print(book_publisher)
```

Издательство Penguin Books выпустило больше всего книг (`42`).

## Определите автора с самой высокой средней оценкой книг.  с 50+ оценками

```python
best_author_ever_query = """
SELECT
      SUBQ.author AS author,
      AVG(SUBQ.avg_rating) AS avg_rating
FROM
(SELECT
     authors.author AS author,
     books.book_id AS id,
     AVG(ratings.rating) AS avg_rating,
     COUNT(ratings.rating) AS cnt_rating
FROM
    authors
INNER JOIN books ON books.author_id = authors.author_id
INNER JOIN ratings ON ratings.book_id = books.book_id
GROUP BY
    author,
    id
    HAVING
    COUNT(ratings.rating) >= 50) AS SUBQ
GROUP BY
    author
ORDER BY
    avg_rating DESC
"""

best_author_ever = run_query(best_author_ever_query)
print(best_author_ever)
```

Cамая высокая средняя оценка у **J.K. Rowling/Mary GrandPré**

## Посчитайте среднее количество текстовых обзоров пользователей, которые поставили более чем по 50 оценок

```python
avg_text_query = """
SELECT ROUND(avg(count)) 
FROM
(SELECT
    COUNT(text)
FROM 
    reviews
FULL OUTER JOIN (
        SELECT username,
               COUNT(rating_id) as cnt
        FROM ratings
        GROUP BY username
    ) AS cnt_rating
ON cnt_rating.username = reviews.username
WHERE cnt > 50
GROUP BY reviews.username
) as count;                                  
"""

avg_text = run_query(avg_text_query)
print(avg_text)
```

Среднее количество обзоров пользователей, которые поставили более чем `50` оценок - `24`

В таблице **best_author_ever** представлены авторы с самой высокой средней оценкой — при этом учитывались только книги с более чем *50* пользовательскими рейтингами, что делает эти оценки наиболее надёжными.

В таблице **ratings_reviews** отражены средние оценки и количество обзоров по книгам — особое внимание стоит обратить на книги с более чем тремя обзорами, так как среди книг с двумя обзорами оценки часто завышены.
